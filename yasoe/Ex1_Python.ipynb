# Ex 9: Computing

Write a program that asks the user for a number of seconds and prints out how many minutes and seconds that is. For instance, 200 seconds is 3 minutes and 20 seconds.

```python
import math
userInput = input("Enter a number of seconds ")
userInputInt = int(userInput)
if userInputInt >= 60 :
  minutes = userInputInt/60
  seconds = userInputInt % 60

print("{} minutes and {} seconds".format(math.floor(minutes),seconds))
```

# Ex 10: Using Math Module

Write a program that asks the user for a number and then prints out the sine, cosine, and tangent of that number.

```python
import math
num = float(input("Enter a number:"))
sine_value = math.sin(int(num))
cosine_value = math.cos(int(num))
tangent_value = sine_value/cosine_value
print("sine {}, cosine {}, tangent {}".format(sine_value,cosine_value,tangent_value))
```
