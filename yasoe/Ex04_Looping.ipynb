Ex 1: Write a program that uses a while loop (not a for loop) to read through a string and print
the characters of the string one-by-one on separate lines.

```python
userInput = input("Enter any String")
lengthOfString = len(userInput)
startingPoint = 0
while startingPoint < lengthOfString :
    print(userInput[startingPoint])
    startingPoint = startingPoint +1
```

Ex 2: A good program will make sure that the data its users enter is valid. Write a program that
asks the user for a weight and converts it from kilograms to pounds. Whenever the user
enters a weight below 0, the program should tell them that their entry is invalid and then ask
them again to enter a weight. [Hint: Use a while loop, not an if statement].

```python
userInput = input("Enter a weight in Kilo")
userInputFloat = float(userInput)
retries = 0
show = True

while userInputFloat < 0 :
    if retries >= 3 :
        show = False
        break
    print("entry is invalid")
    userInput = input("Enter a weight in Kilo")
    userInputFloat = float(userInput)
    retries = retries + 1
    show = True
    
if show :
    print('{} Kilo is in {} Pounds'.format(userInputFloat,userInputFloat * 2.20462))
```

Ex 3: Write a program that allows the user to enter any number of test scores. The user indicates
they are done by entering in a negative number. Print how many of the scores are A’s (90 or
above). Also print out the average

```python
userInput = input("Enter scores")
userInputInt = int(userInput)
userInputList = []

while userInputInt > 0 :
    userInputList.append(userInputInt)
    
    userInput = input("Enter scores")
    userInputInt = int(userInput)

result = 0
ninetyAndAbove = 0
for score in userInputList :

    if score >= 90 :
        ninetyAndAbove = ninetyAndAbove +1
        
    result = result + score

print('Result is {}'.format(result))
print('Length is {}'.format(len(userInputList)))
print('No of Score above 90 is {}'.format(ninetyAndAbove))
print('The Average is {}'.format(result/len(userInputList)))
```

Ex 4: Write a program that asks the user for a string and a letter. Using a for loop, the
program should print the index of the first occurrence of that letter and a message if the
string does not contain the letter.

```python
userInput = input("Enter a letter")
aSearchCharacter = input("Enter a search character(to search from the above string)")

if aSearchCharacter in userInput :
    print("found")
    for index, character in enumerate(userInput) :
        print(index, character)
        if aSearchCharacter == character :
            print("Index of the search character is at {}".format(index))
            break
else:
    print("Not Found")
```
