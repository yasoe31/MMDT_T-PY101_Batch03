Ex 3: Write a function called number_of_factors that takes an integer and returns how many
factors the number has.

```python
def number_of_factors(givenInt) :
    returnValue = []
    for i in range(1,givenInt+1) :
        if givenInt % i == 0 :
            returnValue.append(i)
    return returnValue

number_of_factors(36)
    
```

Ex4: Write a function called first_diff that is given two strings and returns the first location in
which the strings differ. If the strings are identical, it should return -1.Ignore the case in your function. It means h and H are the same.

```python
def first_diff(first,second):
    fLowered = first.lower()
    sLowered = second.lower()
    
    if fLowered == sLowered :
        return -1

    lengthToUse = len(first) if len(first) < len(second) else len(second)

    #print(lengthToUse)

    for idx in range(lengthToUse) :

        if fLowered[idx] != sLowered[idx] :
            return idx

    return lengthToUse

    
#first_diff("abc","acb")
#first_diff("abc","abcd")     
#first_diff("abc","eabcd")   
first_diff("abcde","abcd")         
    
        
```

Ex7:  Write a function called change_case that given a string, returns a string with each upper  case letter replaced by a lower case letter and vice-versa

```python
def change_case(givenString) :
    returnString = ""
    for char in givenString :
        print(char)
        if char.islower() :
            print(char +'is lower')
            returnString = returnString + char.upper()
        else :
            print(char +'is upper')
            returnString = returnString + char.lower()

    return returnString

change_case("aBcdE")
            
            
```

Ex8:  Write a function called primes that is given a number n and returns a list of the first n  primes. Let the default value of n be 100.

```python
import math

def primes(n = 100) :
    primeList = []
    if n > 1:

        for i in range(2, n + 1):
            if i == 2 or i==3 or i == 5 or i == 7:
                primeList.append(i)

            elif i % 2 != 0 and i % 3 !=0 and i % 5 !=0 and i % 7 !=0 :
                primeList.append(i)
    

    return primeList

#primes(9)
primes()
```

Ex9: A Tic-tac-toe board can be represented be a 3×3 two-dimensional list, where zeroes stand for  empty cells, ones stand for X’s and twos stand for O’s. Write a function that is given such a list and randomly chooses a spot in which to place  a 2 (O). The spot chosen must currently be empty and a spot must be chosen.

```python
import random

def tictactoe(board) :
    print(board)
    hasNotChosen = True
    counter = 0

    while hasNotChosen :
        
        localRandomRow = random.randrange(0, 3)
        print(localRandomRow)
        localRandomCol = random.randrange(0, 3)
        print(localRandomCol)

        if board [localRandomRow] [localRandomCol] == 0 :
            print("Chosen spot is at row : {}, col : {}".format(localRandomRow,localRandomCol))
            board[localRandomRow][localRandomCol] = 2
            print(board)

            hasNotChosen = False

        counter = counter + 1
        if counter == 9 :
            print("killed infinite loop")
            hasNotChosen = False # kill infinite loop
    

#tictactoe([[1,2,3],[4,5,6],[7,8,9]])
tictactoe([[0,0,1],[2,0,1],[0,1,2]])
```
